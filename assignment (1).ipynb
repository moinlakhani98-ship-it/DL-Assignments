```python
y = df['final_grade']
X = df.drop(columns=['final_grade', 'pass_fail', 'grade_category', 'student_id'])
print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
income_map = {'Low': 0, 'Medium': 1, 'High': 2}
X['family_income'] = X['family_income'].map(income_map)

edu_map = {'High School': 0, 'Bachelor': 1, 'Master': 2, 'PhD': 3}
X['parent_education'] = X['parent_education'].map(edu_map)

# Binary columns -> 0/1 mapping
X['internet_access'] = X['internet_access'].map({'No': 0, 'Yes': 1})
X['school_type'] = X['school_type'].map({'Public': 0, 'Private': 1})

# Nominal columns (no order) -> One-Hot Encoding
X = pd.get_dummies(X, columns=['gender', 'device_type', 'extracurriculars'], drop_first=True)

print("\nX shape after encoding:", X.shape)
print(X.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numeric_cols = [
    'age', 'study_hours', 'attendance', 'sleep_hours', 'previous_grade',
    'assignments_completed', 'practice_tests_taken', 'group_study_hours',
    'notes_quality_score', 'time_management_score', 'motivation_level',
    'mental_health_score', 'screen_time', 'social_media_hours',
    'family_income', 'parent_education'
]

scaler =    StandardScaler()

# Fit sirf training data par, test data par sirf transform (data leakage avoid karne ke liye)
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])

print("\n--- Scaled Training Data (Sample) ---")
print(X_train[numeric_cols].head())

print("\nPreprocessing Complete!")
print("Final X_train shape:", X_train.shape)
print("Final X_test shape:", X_test.shape)
```
